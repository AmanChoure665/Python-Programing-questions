Looping Through a List

```python
lst = [1,32,5,3,56,6,3636,7,6636,363636,55,676,76,7,7,4,8,5]

n = len(lst)
i = 0
count = 0
while i <= n - 1:
    if lst[i] % 2 == 0:
        print(lst[i], end = " ")
        count += 1
    i += 1
    
print(f"Even count = {count}")
```

```python
# from 1 to 100, make a list of all prime number
def is_prime(n):
    factors = 0
    for i in range(1,n+11):
        if n % i == 0:
            factors += 1
    if factors > 2:
        return False
    return True


new_lst = [i for i in range(2,101) if is_prime(i)]

print(new_lst)
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9],[5,5,5]]


for i in matrix:
    for j in i:
        print(j, end=" ")
    print()

rows = len(matrix)
columns = len(matrix[0])
total = 0
for i in range(0,rows):
    for j in range(0,columns):
        total += matrix[i][j]

print(total, end=" ")

    
```

```python
"""
Q. Given a 3x3 matrix, print only the main diagonal elements and replace
everything else with an asterisk (*).
# 1 2 3
# 4 5 6
# 7 8 9

# Expected Output:
# * * 1
# * 5 *
# 9 * *

"""

matrix = [
    [1, 2, 3, 5],
    [4, 5, 6, 1],
    [7, 8, 9, 7],
    [9, 7, 3, 1],
]


def lower_triangle(matrix):
    rows = len(matrix)
    col = len(matrix[0])
    for i in range(0,rows):
        for j in range(0,col):
            if i+j == 3:
                print(matrix[i][j],end=" ")
            else:
                print("*",end=" ")
        print()

lower_triangle(matrix)
```

```python
# Make a function which return min and max of a list 

def min_max(lst):
    mini = min(lst)
    maxi = max(lst)
    return maxi, mini

ans1, ans2 = min_max([2,5,6,55,7,85,7,33,46,7])

print(f"Maximum number = {ans1}")
print(f"Minimum number = {ans2}")
```
