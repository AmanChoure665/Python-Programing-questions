```python
num = int(input())
for i in range(1,num+1):
    for k in range(1,num-i+1):
        print(" ",end=" ")
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
num = int(input())
for i in range(num,0,-1):
    for k in range(1,i-1+1):
        print(" ",end=" ")
    for j in range(num,i-1,-1):
        print(j,end=" ")
    print()
```

```python
def vote(age):
    
    if age >= 18:
        return True
    else:
        return False

ans = vote(int(input("Enter age: ")))

print(ans)
```

```python
def prime(num):
    fac = 0
    for i in range(1,num+1):
        if num % i == 0:
            fac += 1
    print(fac)

    if fac == 2:
        return True
    return False

ans = prime(int(input("Enter a number: ")))

print(ans)    
```

```python
#Global variable

count = 0

def increase():
    global count
    count += 1
    # count = 1
    print(f"Inside function count = {count}")

increase()
increase()
increase()
print(f"Outside function count = {count}")
```

# Lambda function

```python
is_adult = lambda age: True if age >= 18 else False

print(is_adult(int(input("Enter age: "))))
```

```python
#return cube of a number
cube = lambda num: num **3
print(cube(3))
```

```python
#return number is mossitive or negative
number = lambda num: "positive" if num >= 0 else "negative"

print(number(3))
print(number(-2))
```

List Indexing 

```python
fruits = ["mango",'apple',"banana","orange","grapes"]

print(fruits[0])
print(fruits[-4])
print(fruits[4])
print(fruits[3])
print(fruits[-1]) 

n = len(fruits)
print(f"last element = {fruits[n-1]}")

fruits[2] = 'kiwi'
print(fruits)
```

List Slicing

```python
lst = [1,32,5,3,56,6,3636,7,6636,363636,55,676,76,7,7,4,8,5]

print(lst[0:4])
print(lst[4:9])
print(lst[1:5])
print(lst[-1:-13:-1])
print(lst[-6:-2])
print(lst[::2])
print(lst[::-1])

print(sorted(lst))
```

Looping Through a List

```python
lst = [1,32,5,3,56,6,3636,7,6636,363636,55,676,76,7,7,4,8,5]

n = len(lst)
i = 0
count = 0
while i <= n - 1:
    if lst[i] % 2 == 0:
        print(lst[i], end = " ")
        count += 1
    i += 1
    
print(f"Even count = {count}")
```

```python
lst = [1,32,5,3,56,6,3636,7,6636,363636,55,676,76,7,7,4,8,5]

n = len(lst)
i = n -1

while i >= 0:
    print(lst[i],end = " ")
    i -= 1
```

```python
# Using for loop

lst = [1,32,5,3,56,6,3636,7,6636,363636,55,676,76,7,7,4,8,5]

n = len(lst)
total = 0

for i in lst:
    print(i,end=" ")
    total += i

print(f"Sum of list {total}")
```
