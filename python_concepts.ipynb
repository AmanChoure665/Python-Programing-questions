Escape Sequences:

\n - New Line
\t - Tab space
\\ - Backslash
\" - double quotes
\' - Single quotes

```python
print("Hello\nworld")
print("Hello\tworld")
print("Hello\\world")
print("Hello\"world\"")
print("Hello\'world")
```

#Pattern Printing


```python
# Q1
num1 = int(input())

for i in range(1,num1+1):
    for j in range(1,num1+1):
        print(j, end=" ")
    print()
```

```python
# q2
num1 = int(input())

for i in range(1,num1+1):
    for j in range(1,num1+1):
        print(i, end=" ")
    print()
```

```python
# Q3
num1 = int(input())

for i in range(num1,0,-1):
    for j in range(1,num1+1):
        print(i, end=" ")
    print()
```

```python
num = int(input())
for i in range(1,num+1):
    for k in range(1,num-i+1):
        print(" ",end=" ")
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
num = int(input())
for i in range(num,0,-1):
    for k in range(1,i-1+1):
        print(" ",end=" ")
    for j in range(num,i-1,-1):
        print(j,end=" ")
    print()
```

```python
def vote(age):
    
    if age >= 18:
        return True
    else:
        return False

ans = vote(int(input("Enter age: ")))

print(ans)
```

```python
def prime(num):
    fac = 0
    for i in range(1,num+1):
        if num % i == 0:
            fac += 1
    print(fac)

    if fac == 2:
        return True
    return False

ans = prime(int(input("Enter a number: ")))

print(ans)    
```
