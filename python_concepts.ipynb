```python
# from 1 to 100, make a list of all prime number
def is_prime(n):
    factors = 0
    for i in range(1,n+11):
        if n % i == 0:
            factors += 1
    if factors > 2:
        return False
    return True


new_lst = [i for i in range(2,101) if is_prime(i)]

print(new_lst)
```

```python
"""
Q. Given a 3x3 matrix, print only the main diagonal elements and replace
everything else with an asterisk (*).
# 1 2 3
# 4 5 6
# 7 8 9

# Expected Output:
# * * 1
# * 5 *
# 9 * *

"""

matrix = [
    [1, 2, 3, 5],
    [4, 5, 6, 1],
    [7, 8, 9, 7],
    [9, 7, 3, 1],
]


def lower_triangle(matrix):
    rows = len(matrix)
    col = len(matrix[0])
    for i in range(0,rows):
        for j in range(0,col):
            if i+j == 3:
                print(matrix[i][j],end=" ")
            else:
                print("*",end=" ")
        print()

lower_triangle(matrix)
```

Adding and removing element in sets

```python
''' Adding Elements '''

pythonfruits = {"apple","mango"}

# add() - adds a single element
pythonfruits.add("banana")
print(pythonfruits)


# update() - adds multiple elements
pythonfruits.update(["kiwi","graps"])
print(pythonfruits)

# adding a duplicate is silently ignored
pythonfruits.add("banana")
print(pythonfruits)
```

```python
''' Removing Elements '''

pythonfruits = {'kiwi', 'banana', 'graps', 'apple', 'mango'}

# remove() - gives error if key dosent exist
pythonfruits.remove("banana")
print(pythonfruits)


# discard() - do nothing if key dosent exist
pythonfruits.discard("applee")
print(pythonfruits)


# pop() - removes random element
removed = pythonfruits.pop()
print(removed)


# clear() - print empty set
pythonfruits.clear()
print(pythonfruits)
```
