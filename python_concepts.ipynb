Escape Sequences:

\n - New Line
\t - Tab space
\\ - Backslash
\" - double quotes
\' - Single quotes

```python
print("Hello\nworld")
print("Hello\tworld")
print("Hello\\world")
print("Hello\"world\"")
print("Hello\'world")
```

 operator precedence

from left to write >>>>
() > ** > * > / > // > % > + > -

```python
age = int(input())

print("Adult" if age >=18 else "Not Adult")
```
